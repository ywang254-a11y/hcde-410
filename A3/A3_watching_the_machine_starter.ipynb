What percentage of valid responses falls into each answer category? Treat capitalization and surrounding whitespace as irrelevant, and exclude missing responses from the percentage denominator. 

Report the count and percentage for each category, the number of valid and missing responses, and briefly explain how you arrived at your answer. Round percentages to two decimal places.

```python
import pandas as pd

responses = [
    "Yes", "no", "No", "Yes", "Maybe",
    "", "yes", "no", "YES", "yes",
    "NO", " ", "Yes", "No", "no",
    "yes", "maybe", "NO", "yes", "Yes"
]

df = pd.DataFrame({"response": responses})

df["cleaned"] = df["response"].str.strip().str.lower()
df["cleaned"] = df["cleaned"].replace("", pd.NA)

valid = df["cleaned"].dropna()

counts = valid.value_counts()

summary = pd.DataFrame({
    "Count": counts,
    "Percentage": counts / len(valid) * 100
})

print("Total:", len(df))
print("Valid:", len(valid))
print("Missing:", df["cleaned"].isna().sum())

display(summary.round(2))
```

**What varied, and what held up?** *Summarize the divergence across runs and models, and how you verified each answer against ground truth.*

Across all four runs, I used ChatGPT three times and Claude once, and all four runs produced the same correct results: 9 Yes responses (50.00%), 7 No responses (38.89%), and 2 Maybe responses(11.11%), with 18 valid responses and 2 missing values. However, the structure and level of detail varied between runs. Run 1 provided the most detailed response, including a table, chart, formulas, and explanations. Runs 2 and 3 were more concise but included additional usability interpretations, suggesting that the draft-saving feature might present challenges for some participants. Run 4 focused more on the data cleaning and counting process by listing the specific response numbers in each category, making the results easier to verify.

To verify the outputs, I used pandas to independently clean the original 20 responses, normalize capitalization and whitespace, exclude missing values, and calculate the counts and percentages. I compared these results with each of the four model outputs and confirmed that all were correct. This result showed that although the numerical results remained consistent, the models differed in how they presented and interpreted the same data. In particular, the additional interpretations in Runs 2 and 3 went beyond what the synthetic survey data alone could establish, reminding me that accurate calculations do not necessarily mean every interpretation is supported by evidence.
